```python
from huggingface_hub import InferenceClient
from google.colab import userdata
client = InferenceClient(api_key=userdata.get('assignment_key'))
user_input = input("Ask the three things to the person:\n\n")

print("\nAsk the questions as below:")
print("1. What is your available space? ")
print("2. How much sunlight does the area get?")
print("3. What is your gardening experience level?")

choice = input("\nEnter your choice (1-3)\n\n")

if choice == "1":

 instruction = (
      "You are a sustainable gardening expert. Design an application that adapts the dwellers to transform small spaces in to thriving green oases. "
      "Rewrite the available spaces around your surroundings. "
 )

elif choice == "2":
    instruction = (

      "You are a sustainable gardening expert. Design an application that adapts the dwellers to transform small spaces in to thriving green oases. "
      "Rewrite the available sunlight coverage around your surroundings. "


        )

elif choice == "3":
      instruction = (

      "You are a sustainable gardening expert. Design an application that adapts the dwellers to transform small spaces in to thriving green oases. "
      "Rewrite garderning experiences. "


      )


else:
     print("\nInvalid choice")

#print("Choice =", choice)
#print("Instruction =")
#print(instruction)

prompt = f"{instruction}\n\nYour surrundings info:\n{user_input}"

messages = [
{
"role": "user",
"content": prompt
}

]

completion = client.chat.completions.create(
model="meta-llama/Llama-3.1-8B-Instruct",
messages=messages
)
print("\n")
print(completion.choices[0].message.content)
```
