```python
from huggingface_hub import InferenceClient
from google.colab import userdata
client = InferenceClient(api_key=userdata.get('assignment_key'))
user_input = input("Enter the handyman service description:\n\n")
print("\nSelect Target Audience:")
print("1. The Busy Professional")
print("2. Budget-Conscious Homeowner")
print("3. The Senior Citizen")
print("4. The Property Manager")
choice = input("\nEnter your choice (1-4)\n\n")
if choice == "1":
    instruction = (
        "You are a seasoned digital marketing strategist, audience segmentation expert, and professional marketing copywriter. "
        "Rewrite the handyman service description for a Busy Professional. "
        "Focus on efficiency, timesaving, reliability, 'set it and forget it'. "
        "Write one persuasive marketing paragraph with 60-80 words"
        "Do not provide analysis. "
        "Return only rewritten service description, change the tone, emphasis, and hook to match theaudience. "
    )
elif choice == "2":
    instruction = (
        "You are a seasoned digital marketing strategist, audience segmentation expert, andprofessional marketing copywriter. "
        "Rewrite the handyman service description for a Budget-Conscious Homeowner."
        "Focus on affordability, value for money, and preventing costly future damage. "
        "Write one persuasive marketing paragraph with 60-80 words. "
        "Do not provide analysis"
        "Return only rewritten service description, change the tone, emphasis, and hook to matchthe audience. "
    )
elif choice == "3":
    instruction = (
        "You are a seasoned digital marketing strategist, audience segmentation expert, andprofessional marketing copywriter. "
        "Rewrite the handyman service description for The Senior Citizen."
        "Focus on trust, safety, friendliness, meticulousness, and respect."
        "Write one persuasive marketing paragraph with 60-80 words. "
        "Do not provide analysis. "
        "Return only rewritten service description, change the tone, emphasis, and hook to match the audience. "
    )
elif choice == "4":
    instruction = (
        "You are a seasoned digital marketing strategist, audience segmentation expert, andprofessional marketing copywriter. "
        "Rewrite the handyman service description for the Property Manager."
        "Focus on professionalism, quick turnaround, high-quality finishing, and long-termproperty value."
        "Write one persuasive marketing paragraph with 60-80 words. "
        "Do not provide analysis. "
        "Return only rewritten service description, change the tone, emphasis, and hook to match the audience. "
    )
else:
    print("\nInvalid choice")
#print("Choice =", choice)
#print("Instruction =")
#print(instruction)
prompt = f"{instruction}\n\nService Description:\n{user_input}"
messages = [
    {
        "role": "user",
        "content": prompt
    }
]
completion = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=messages
)
print("\n")
print(completion.choices[0].message.content)
```
